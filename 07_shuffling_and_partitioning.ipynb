````python
# [+] PySpark 시작
from pyspark import SparkConf, SparkContext
conf = SparkConf().setMaster('local').setAppName('rdd-transformations-and-actions')
sc = SparkContext(conf=conf)
sc.setLogLevel("ERROR")  # 로그 레벨 설정
````

````python
# 데이터셋 경로 및 파일 이름
path = "C:/Dev/bigdata-spark-lab/data"
filename = "fhvhv_tripdata_2020-03_short.csv"
````

````python
# [+] 데이터셋 로딩 및 헤더 추출
lines = sc.textFile(path + filename)
header = lines.first()
````

````python
# [+] 헤더 제거
filtered_lines = lines.filter(lambda line: line != header)
filtered_lines.take(5)
````

*filtered_lines 데이터*

````python
# [+] 승차일자 추출하기(K-V RDD로 저장): map()
dates = filtered_lines.map(lambda line: line.split(',')[1])
dates.take(5)
````

### **일별 승차횟수 계산**
---

#### **reduceByKey( )**

````python
# [+] reduceByKey()를 이용한 일별 승차횟수 계산
reduced = dates.map(lambda date: (date, 1)).reduceByKey(lambda a, b: a + b)
````

```%%time```: 셀 단위 실행 시간 측정(셀 최상단에 작성)



````python
%%time
# 처리 시간 측정
reduced.collect()
````

#### **groupByKey( )**

````python
# [+] groupByKey()를 이용한 일별 데이터 그룹핑
groups = dates.groupByKey()
````

````python
# [+] mapValues()를 이용한 일별 승차횟수 계산
counts = groups.mapValues(len)
````

```%time```: 문장 단위 실행 시간 측정(문장 앞에 작성)

````python
# [+] 처리 시간 측정
%time counts.collect()
````

### **RDD 파티셔닝(Partitioning)**
---

````python
# 데이터셋 불러오기
path = 'C:/Dev/bigdata-spark-lab/data'
filename = 'id_and_name.csv'
lines = sc.textFile(path + filename)
````

````python
# [+] RDD 값 10개 출력
lines.take(10)
````

````python
# [+] 헤더 제거
header = lines.first()
filtered_lines = lines.filter(lambda line: line != header)
````

````python
# 파티션 수 설정
n = 5
````

#### **K-V RDD로 변환**
*filtered_lines*
```
['201958114,최창성',
 '202058062,유예지',
 '202122035,이원모',
 '202158050,맹서희',
 '202158083,신동민',
 ...
```

*kv_lines*
```
[['201958114', '최창성'],
 ['202058062', '유예지'],
 ['202122035', '이원모'],
 ['202158050', '맹서희'],
 ['202158083', '신동민'],
 ...
```


````python
# [+] K-V RDD로 변환
kv_lines = filtered_lines.map(lambda x:x.split(','))
kv_lines.take(5)
````

#### 3.2 Hash Partitioning
key(x)에 대한 해시 함수를 정의하여 파티셔닝을 수행

````python
# [+] 모듈로 해시 함수를 이용한 파티셔닝: partitionBy()
parted = kv_lines.partitionBy(n, lambda x: hash(x) % n)
````

````python
# [+] 파티션별 RDD 값 출력
for i in range(n):
    print(f"Partition {i}: {parted.glom().collect()[i]}")
````

````python
# [+] 파티션 수 조정: repartition()
reparted = kv_lines.repartition(n)
````

````python
# [+] 파티션별 RDD 값 출력
reparted.glom().collect()
````

````python
# 파티션 수 조정: coalesce()
combined = reparted.coalesce(numPartitions=3)
````

````python
# 파티션별 RDD 값 출력
combined.glom().collect()
````

#### **Range Partitioning**
+ Range partitioning 수행을 위해 RDD 객체를 DataFrame으로 변환
+ DataFrame 생성을 위해 SparkSession 사용

````python
# SparkSession 임포트 및 객체 생성
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("RepartitionByRangeExample").getOrCreate()
````

````python
# DataFrame 생성
df = spark.createDataFrame(parted, ["id", "name"])
````

````python
# DataFrame 출력
df.show()
````

````python
# Range Partitioning: repartitionByRange()
reparted = df.repartitionByRange(3, 'id')
````

파티션별 RDD 값 출력
+ ```rdd```: DataFrame에 포함된 RDD 객체
+ ```glom()```: 파티션별로 RDD값 구성

````python
# .rdd: DataFrame의 RDD 객체 반환
reparted.rdd.glom().collect()
````
